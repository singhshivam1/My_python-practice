# Find the Fabanoci series less than 50. 
# Find the even number from series which are less than 50 and provide the sum

```python
index = 1
next = 2
#print(index)
#print(next)
even_sum = 2
while (index + next) < 50:
    
    sum = index + next
 #   print(sum)
    if sum % 2 == 0:
        even_sum += sum  
    index = next
    next = sum
    
print("Sum of even-valued terms in the Fibonacci sequence:", even_sum)
```
